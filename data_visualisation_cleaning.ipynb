```python
pd.set_option("display.max_rows", 300)

straftaten = pd.Series(df["Straftat"].unique(), name="Straftat")
straftaten
```

```python
auswahl = [
    "Unrechtmässige Aneignung (Art. 137)",
    "Veruntreuung (Art. 138)",
    "Diebstahl (Art. 139)",
    "Einbruchdiebstahl (Art. 139)",
]

entwicklung = df[
    df["Straftat"].isin(auswahl)
    & (df["Kanton"] == "Schweiz")
    & (df["Aufenthaltsgruppe"] == "Total alle Beschuldigte")
    & (df["Geschlecht"] == "Geschlecht - Total")
    & (df["Altersklasse"] == "Altersklasse - Total")
]

tabelle = entwicklung.pivot_table(index="Jahr", columns="Straftat", values="DATA", observed=True)
display(tabelle)

tabelle.plot(
    figsize=(10, 5),
    marker="o",
    grid=True,
    title="Polizeilich registrierte Beschuldigte, ganze Schweiz",
    ylabel="Anzahl Beschuldigte",
)
```
