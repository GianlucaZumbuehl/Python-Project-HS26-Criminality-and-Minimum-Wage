```python
pd.set_option("display.max_rows", 300)

straftaten = pd.Series(df["Straftat"].unique(), name="Straftat")
straftaten
```

```python
auswahl = [
    "Unrechtmässige Aneignung (Art. 137)",
    "Veruntreuung (Art. 138)",
    "Diebstahl (Art. 139)",
    "Einbruchdiebstahl (Art. 139)",
]

entwicklung = df[
    df["Straftat"].isin(auswahl)
    & (df["Kanton"] == "Schweiz")
    & (df["Aufenthaltsgruppe"] == "Total alle Beschuldigte")
    & (df["Geschlecht"] == "Geschlecht - Total")
    & (df["Altersklasse"] == "Altersklasse - Total")
]

tabelle = entwicklung.pivot_table(index="Jahr", columns="Straftat", values="DATA", observed=True)
display(tabelle)

tabelle.plot(
    figsize=(10, 5),
    marker="o",
    grid=True,
    title="Polizeilich registrierte Beschuldigte, ganze Schweiz",
    ylabel="Anzahl Beschuldigte",
)
```

```python
import matplotlib.pyplot as plt
d = df[(df["Straftat"] == "Diebstahl (Art. 139)") & (df["Kanton"] != "Schweiz") & (df["Aufenthaltsgruppe"] == "Total alle Beschuldigte") & (df["Geschlecht"] == "Geschlecht - Total") & (df["Altersklasse"] == "Altersklasse - Total")]
p = d.pivot_table(index="Jahr", columns="Kanton", values="DATA"); p = p / p.loc[2016] * 100
ax = p.drop(columns="Neuchâtel").plot(color="lightgrey", lw=1, legend=False, figsize=(10, 5), grid=True)
p.drop(columns=["Neuchâtel", "Jura", "Genève", "Ticino", "Basel-Stadt"]).mean(axis=1).plot(ax=ax, color="black", ls="--", label="Ø Kantone ohne Mindestlohn")
p["Neuchâtel"].plot(ax=ax, color="red", lw=2.5, marker="o", label="Neuchâtel")
ax.legend(handles=ax.lines[-2:]); ax.axvline(2017, color="grey", ls=":"); ax.set(title="Beschuldigte Diebstahl (Art. 139), Index 2016 = 100", ylabel="Index"); plt.show()
```

```python
import matplotlib.pyplot as plt
mw = {"Neuchâtel": 2017, "Jura": 2018, "Genève": 2020, "Ticino": 2021, "Basel-Stadt": 2022}
d = df[(df["Straftat"] == "Diebstahl (Art. 139)") & (df["Kanton"] != "Schweiz") & (df["Aufenthaltsgruppe"] == "Total alle Beschuldigte") & (df["Geschlecht"] == "Geschlecht - Total") & (df["Altersklasse"] == "Altersklasse - Total")]
p = d.pivot_table(index="Jahr", columns="Kanton", values="DATA")
fig, axes = plt.subplots(2, 3, figsize=(18, 9)); axes.flat[-1].axis("off")
for ax, (k, j) in zip(axes.flat, mw.items()):
    q = p / p.loc[j - 1] * 100
    q.drop(columns=k).plot(ax=ax, color="lightgrey", lw=1, legend=False)
    q.drop(columns=list(mw)).mean(axis=1).plot(ax=ax, color="black", ls="--", label="Ø Kantone ohne Mindestlohn")
    q[k].plot(ax=ax, color="red", lw=2.5, marker="o", label=k)
    ax.legend(handles=ax.lines[-2:]); ax.axvline(j, color="grey", ls=":"); ax.set(title=f"{k} (Mindestlohn ab {j})", ylabel=f"Index {j - 1} = 100", ylim=(0, 250)); ax.grid(True)
fig.suptitle("Beschuldigte Diebstahl (Art. 139)"); plt.tight_layout(); plt.show()
```

```python
import numpy as np, matplotlib.pyplot as plt
mw = {"Neuchâtel": 2017, "Jura": 2018, "Genève": 2020, "Ticino": 2021, "Basel-Stadt": 2022}
d = df[(df["Straftat"] == "Diebstahl (Art. 139)") & (df["Kanton"] != "Schweiz") & (df["Aufenthaltsgruppe"] == "Total alle Beschuldigte") & (df["Geschlecht"] == "Geschlecht - Total") & (df["Altersklasse"] == "Altersklasse - Total")]
p = d.pivot_table(index="Jahr", columns="Kanton", values="DATA"); kontrolle = np.log(p.drop(columns=list(mw)).sum(axis=1))
fig, ax = plt.subplots(figsize=(10, 5))
for k, j in mw.items():
    diff = np.log(p[k]) - kontrolle; diff = diff - diff[diff.index < j].mean()
    ax.plot(diff.index - j, diff, marker="o", label=k)
ax.axhline(0, color="black", lw=1); ax.axvline(-0.5, color="grey", ls=":"); ax.grid(True); ax.legend()
ax.set(title="Diebstahl: Abstand zur Kontrollgruppe (Kantone ohne Mindestlohn)", xlabel="Jahre seit Einführung Mindestlohn", ylabel="Log-Abstand (≈ %)"); plt.show()
```
